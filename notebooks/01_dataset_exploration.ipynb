Inspect the downloaded corpus and validate split integrity. No synthetic benchmark records.

```python
from pathlib import Path
import sys

if Path.cwd().name == "notebooks":
    import os

    os.chdir("..")
sys.path.insert(0, str(Path.cwd()))
from ml.datasets.loader import load_corpus, load_judgments

papers, stats = load_corpus("data/scifact")
stats
```

```python
train, _ = load_judgments("data/scifact", papers, "train")
test, qrels = load_judgments("data/scifact", papers, "test")
assert not set(train) & set(test)
len(train), len(test), len(papers)
```

```python
import matplotlib.pyplot as plt

plt.hist([len(p.text().split()) for p in papers], bins=40)
plt.xlabel("Whitespace tokens, title plus abstract")
plt.ylabel("Documents")
```
